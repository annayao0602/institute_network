```python
import pandas as pd

full = pd.read_excel("UVAdepartments.xlsx")
cleaned = pd.read_excel("null_cleaned.xlsx")

full.columns = full.columns.str.strip()
cleaned.columns = cleaned.columns.str.strip()

keys = ["_faculty_key", "_department_key"]

for df in [full, cleaned]:
    df["_faculty_key"] = (
        df["Faculty Name"].astype("string").str.strip().str.casefold()
    )
    df["_department_key"] = (
        df["Department"].astype("string").str.strip().str.casefold()
    )

if cleaned.duplicated(keys).any():
    raise ValueError(
        "The cleaned file has duplicate Faculty Name + Department pairs."
    )

if full[keys].isna().any().any() or cleaned[keys].isna().any().any():
    raise ValueError("Faculty Name and Department must not be missing.")

# Include all columns from both files.
columns = list(dict.fromkeys(
    [c for c in full.columns if c not in keys]
    + [c for c in cleaned.columns if c not in keys]
))

full_indexed = full.set_index(keys).reindex(columns=columns)
cleaned_indexed = cleaned.set_index(keys).reindex(columns=columns)

# Align cleaned rows to the full list's original row order.
matched = full_indexed.index.isin(cleaned_indexed.index)
aligned_cleaned = cleaned_indexed.reindex(full_indexed.index)

result = full_indexed.copy()

# Replace entire matched rows, including blank values.
result.loc[matched, :] = aligned_cleaned.loc[matched, :].to_numpy()

result = result.reset_index(drop=True)

assert len(result) == len(full)
#drop "Other Identifiers" column and "Other Identifier Type" column
result = result.drop(columns=["Other Identifiers", "Other Identifier Type"])

#change column School of Education & Human Developement to School of Education and Human Development
result["School"] = result["School"].replace({"School of Education & Human Development": "School of Education and Human Development"})

result.to_excel("UVA_departments_updated.xlsx", index=False)

print(f"Replaced {matched.sum():,} rows; saved {len(result):,} total rows.")
```
